```python
class MobilePhone:
    def __init__(self,b,p):
        self.brand = b
        self.price = p
    
    def call(self):
        print(self.brand,"is making a call")

    def show_price(self):
        print("Price :",self.price)
    


phone1 = MobilePhone("Samsung",30000)
phone2 = MobilePhone("Nokia",100000)

phone1.call()
phone2.show_price()
```

In this the __init(self,b,p):

self refernce to the current instance of the class
b , p ====> parameters

the values passed during a  function call to the  respective function

self.brand will put the respective value to the current object 

```python
class BankAccount:
    def __init__(self,name,balance):
        self.name = name
        self.balance = balance

    def deposit(self,amount):
        self.balance += amount
        print(amount,"The amount is deposited")

    def withdraw(self,amount):
        if self.balance < amount:
            print("Insufficient Balance")
        else:
            self.balance -= amount
            print(f"{amount} is withdrawn")
    
    def show_balance(self):
        print(f"The {self.name} holds the account")
        print(f"The balance of {self.name} is {self.balance}")
    


user1 = BankAccount("Joban",400000)
user1.deposit(600000)
user1.withdraw(500)
user1.show_balance()

        
```

```python
class Order:
    def __init__(self,customer_name):
        self.customer_name = customer_name
        self.products = []
        self.total = 0

    def add_products(self,product_name,price):
        self.products.append(product_name)
        self.total += price
    def apply_discount(self,percentage):
        discount = self.total*(percentage/100)
        self.total -= discount
    
    def generate_invoice(self):
        print(f"The customer is {self.customer_name}")
        print(f"The products bought are {self.products}")
        print(f"The final amount to be paid after the discount is {self.total}")


order1 = Order("Joban")
order1.add_products("Laptop",250000)
order1.add_products("Phone",150000)
order1.apply_discount(10)
order1.generate_invoice()

```

```python
#object creation using LookupError

class product:
    def __init__(self,name):
        self.name = name
    
products = ["Laptop","Mobile","Jacket"]
product_objects = []

for p in products:
    obj = product(p)
    product_objects.append(obj)

print(product_objects)
```

```python
class BankAccount:
    def __init__(self,name,balance):
        self.name = name
        self.balance = balance
    
    @classmethod
    def create_savings_account(cls,name):
        return cls(name,5000)
    @classmethod
    def create_current_account(cls,name):
        return cls(name,1000)


acc1 = BankAccount.create_current_account("Ravi")
acc2 = BankAccount.create_savings_account("Shami")
```
