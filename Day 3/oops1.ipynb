```python
class MobilePhone:
    def __init__(self,b,p):
        self.brand = b
        self.price = p
    
    def call(self):
        print(self.brand,"is making a call")

    def show_price(self):
        print("Price :",self.price)
    


phone1 = MobilePhone("Samsung",30000)
phone2 = MobilePhone("Nokia",100000)

phone1.call()
phone2.show_price()
```

In this the __init(self,b,p):

self refernce to the current instance of the class
b , p ====> parameters

the values passed during a  function call to the  respective function

self.brand will put the respective value to the current object
